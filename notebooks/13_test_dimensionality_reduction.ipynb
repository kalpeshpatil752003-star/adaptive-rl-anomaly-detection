```python
from src.preprocessing.loader import DataLoader
from src.preprocessing.cleaner import DataCleaner
from src.preprocessing.encoder import DataEncoder
from src.preprocessing.scaler import DataScaler

from src.feature_engineering.dimensionality_reduction import DimensionalityReducer
```

```python
loader = DataLoader()
cleaner = DataCleaner()
encoder = DataEncoder()
scaler = DataScaler()
```

```python
df = loader.load_csv("industry.csv")

df = cleaner.clean(df)

df = encoder.fit_transform(df)

df = scaler.fit_transform(df)
```

```python
reducer = DimensionalityReducer(
    method="pca",
    n_components=1
)

reduced = reducer.fit_transform(df)

reduced.head()
```
